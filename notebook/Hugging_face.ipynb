``` text 
Part 2: Your First Pipeline
Pipelines are the easiest way to use Hugging Face models. They handle all the complexity for you:
- Downloading the model
- Loading the tokenizer
- Processing your input
- Running inference
- Formatting the output
The question-answering pipeline is crucial for RAG systems. It takes:

A question you want answered
A context containing the information
And it extracts the answer from the context.

This is exactly what happens in RAG: we retrieve relevant documents (context) and then extract answers from them.
```

| -------------------------- | ------------------------------------ |

````python
# Import the tokenizer class.
# A tokenizer converts human text into tokens (numbers)
# that a transformer model can understand.
from transformers import AutoTokenizer
# Import a model specifically designed for Question Answering.
# The model predicts where the answer starts and ends in the context.
from transformers import AutoModelForQuestionAnswering
# Name of the pretrained model stored on Hugging Face.
# This model was trained on the SQuAD question-answering dataset.
model_name = "distilbert-base-cased-distilled-squad"
# Download and load the tokenizer associated with this model.
# Using the matching tokenizer is important because the model
# expects text to be processed in a specific way.
tokenizer = AutoTokenizer.from_pretrained(model_name)
# Download and load the Question Answering model.
# The model weights are obtained from Hugging Face Hub.
model = AutoModelForQuestionAnswering.from_pretrained(model_name)
# Choose GPU if available, otherwise use CPU.
# GPU is much faster for deep learning calculations.
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# Move the model to the selected device.
model.to(device)
# Put the model into evaluation mode.
# This disables training-specific behaviors such as dropout.
model.eval()
# The question we want the model to answer.
question = "Where does the Virgin Mary statue stand?"
# Context containing the information needed to answer the question.
context = "A golden statue of the Virgin Mary stands atop the Main Building's gold dome."
# Convert question and context into model inputs.
inputs = tokenizer(
question,
context,
# Return PyTorch tensors.
# "pt" stands for PyTorch.
return_tensors="pt",
# Keep track of how tokens map back to original text.
# Useful for extracting the final answer span.
return_offsets_mapping=True,
)
print("Loading complete!")
print(model)
````
